# Basic Usage Example

This notebook goes through a minimal example, that represents the core workflows of using the `ProcessPipeline`.
First, we load the `ProcessPipeline` class and python's `pathlib`, which we use to handle file paths.

```python
from pathlib import Path

from image_processing_pipeline import ProcessPipeline
```

Every `ProcessPipeline` requires three informations:
1. The path to a configuration file.
2. The required inputs declared in this config file.
3. A target path for the outputs.

This is a good point to look at the config file **config.yaml** in this directory. You will see, that the config itself defines three groups:
- _Inputs_: All input objects, that need to be available at the start of the pipeline run. Here we need an object `MyNumber`, which we define as `1` below.
- _PipelineSteps_: The `ProcessStep`s that the pipeline will execute in order. This pipeline performs only one step, which
  - will be displayed as "Offset by three" (the `DisplayId` can be any string),
  - will use a `ProcessStep` called `NumberAdder`, which
  - adds all provided numbers in the `Inputs` section,
  - offsetting them by an optional value, defined in the `Options` section, and finally
  - stores them under the name "MyNumberPlus3", defined in the `Deliverable`s section.
  This definition of steps in the pipeline via specificying `DisplayId`, `ProcessStep`, `Inputs`, `Deliverables` and optional `Options` is universal. Every pipeline steps
  will carry these informations.
- _Serialisations_: This section defines, which results will be saved at the end of the pipeline run and where they will be written to. By itself, the pipeline does not
  save any results, only values defined here can be expected to be available at the end of the pipeline run.
  
  Serialisations happens in groups. This helps, for example, if your pipeline does image transformations and data analysis in one go, but you want to keep these results separate.
  Here we only have one group, which stores everythin listed under _Data_ into the specified directory relative to the output directory.

```python
pp = ProcessPipeline(
  inputs={"MyNumber": 1},
  config_path=Path.cwd() / "config.yaml",
  output_dir=Path.cwd() / "outputs",
)
```

The pipeline is now defined and performed a handful of checks, like whether all necessary inputs are available, or whether the config is self-consistent and can produce all results, in requires in other steps.

To run the pipeline now, simply call the `run` method on it.

```python
pp.run()
```
