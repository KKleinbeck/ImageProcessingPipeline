# Image detection example

In this notebook, we will use the process pipeline to apply a basic image processing task: Identify a rectengular cell within an image and extract contents of the cell.

## Data generation
In order to usely test the framework, without the need to rely on actual image data, the framework provides utility methods to generate sample data.
There are a handful of features that allow to costumise the sampla data generation. For more details, inspect `SampleDataConfig`.

```python
from image_processing_pipeline import Visualiser
from image_processing_pipeline.utils.generate_sample_data import SampleDataConfig, generate_sample_frame
```

```python
config = SampleDataConfig()

sample_frame = generate_sample_frame(config)
Visualiser.show_image_stack(sample_frame)
```

We got an image with a handful of structures, a cell, containing blob like shapes, and bit of noise.
Our task will be, to extract the contents of the cell. First though, we serialise the data, such that it can be used as the input to the process pipeline.
For data handling, the framework implements multiple `AbstractProcessData`-based classes, that help with reading and writing data of specific types.
As we want to create a *tiff* image, we use the `ProcessTiffData` realisation.

```python
import numpy as np
from pathlib import Path
from image_processing_pipeline.framework.process_data import ProcessTiffData

data_dir = Path.cwd() / "data"
data_dir.mkdir(exist_ok=True)

ptd = ProcessTiffData(np.reshape(sample_frame, shape=(1, *sample_frame.shape)), "InputImage")
ptd.serialise(data_dir)
```

Now, there exists an *tiff* image under your local *data* directory.

# Extracting the cell

We will go through the config step by step, using the visualiser you have already seen above.
However, for an even deeper look under the hood, you might want to open the process pipeline config and check out the individual process steps in detail.
This will reveal a couple of common usage patterns and best practices.

```python
from image_processing_pipeline import ProcessPipeline
```

```python
pp = ProcessPipeline(
  inputs={"input_image_path": data_dir / "InputImage.tif"},
  config_path=Path.cwd() / "config.yaml",
  output_dir=Path.cwd() / "outputs",
)

pp.run()
```

```python
Visualiser.show_image_stack(pp.data_manager.get("local_variance"))  # ty: ignore[invalid-argument-type]
```
